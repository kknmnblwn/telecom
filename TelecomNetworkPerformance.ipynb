```python
import pandas as pd
```

# Load the .csv file into files

```python
# loading the dataset and initial inspection; getting the file_path from the Files section
file_path = '/content/davao_city_telecom_network_performance_3yr_raw.csv'
df = pd.read_csv(file_path)

df.info()
```

```python
df
```

# Changing log_date col into datetime

```python
df['log_date'] = pd.to_datetime(df['log_date'])
df.info()
```

```python
df
```

# Deletion of rows w/ 'Off-Peak (00:00-17:00)' in time_window col

```python
# deletion of rows containing 'Off-Peak (00:00-17:00)'; NOTE: I put .copy() in order for pandas to make a separate dataset
df = df[df['time_window'] != 'Off-Peak (00:00-17:00)'].copy()
df.info()
```

```python
#checking time_window col

df['time_window']
```

# packet_loss_pct column

```python
df['packet_loss_pct']
```

```python
df['packet_loss_pct'].describe()
```

```python
# replacing -99.00 into its mean value 4.25
df.loc[:, 'packet_loss_pct'] = df['packet_loss_pct'].replace(-99.0, 4.25)
df['packet_loss_pct'].describe()
```

```python
# Display the rows where packet_loss_pct is 0.00
df[df['packet_loss_pct'] == 0.00]
```

```python
df.info()
```

# district & sub_district columns

```python
# Transform 'district' and 'sub_district' columns to title case
df['district'] = df['district'].str.title()
df['sub_district'] = df['sub_district'].str.title()

# Display the unique values of the transformed columns to verify
print("Unique Districts:", df['district'].unique())
print("Unique Sub-Districts:", df['sub_district'].unique())
```

```python
# Checking sub_disctricts according to its district. We can see
df.groupby('district')['sub_district'].unique().to_frame()
```

```python
# Renaming Telomo-Buhangin Proper into Buhangin-Buhangin Proper

df.loc[(df['district'] == 'Talomo') & (df['sub_district'] == 'Buhangin Proper'), 'district'] = 'Buhangin'
df.groupby('district')['sub_district'].unique().to_frame()
```

# average_latency_ms

```python
df['avg_latency_ms'].describe()
```

```python
# 1. Drop the accidentally created row from the index if it exists
if 'avg_latency_ms' in df.index:
    df = df.drop(index='avg_latency_ms')

# 2. Correctly replace the outlier 9999.00 with the mean in-place
df.loc[:, 'avg_latency_ms'] = df['avg_latency_ms'].replace(9999.00, 74.886916)

# 3. Verify the changes
df['avg_latency_ms'].describe()
```
