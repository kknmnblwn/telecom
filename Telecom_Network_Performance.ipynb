```python
import pandas as pd
```

```python
# Load the dataset and inspect its structure and information
file_path = '/content/davao_city_telecom_network_performance_3yr_raw.csv'
df = pd.read_csv(file_path)

# Display summary information about the DataFrame
df.info()
```

```python
#Changing log_date col into datetime format
df['log_date'] = pd.to_datetime(df['log_date'])
df.info()
```

```python
# Option 1: Fill missing ticket counts with 0 (Assuming missing means no tickets were logged)

df['complaint_tickets_logged'] = df['complaint_tickets_logged'].fillna(0)
df.info()
```

```python
df['complaint_tickets_logged'].value_counts()
```

```python
#Changing complaint_tickets_logged col into a integer datatype

df['complaint_tickets_logged'] = df['complaint_tickets_logged'].astype(int)
df.info()
```

```python
#Changing avg_monthly_arpu_php col into float

df['avg_monthly_arpu_php'] = df['avg_monthly_arpu_php'].astype(float)
df.info()
```

```python
# Load the original CSV file to restore only the 'avg_monthly_arpu_php' column
temp_df = pd.read_csv('/content/davao_city_telecom_network_performance_3yr_raw.csv', usecols=['avg_monthly_arpu_php'])
df['avg_monthly_arpu_php'] = temp_df['avg_monthly_arpu_php']

# Confirm that the values have been restored
df.info()
```

```python
#Changing est_ftth_upgrade_cost_per_sub_php col into float

df['est_ftth_upgrade_cost_per_sub_php'] = df['est_ftth_upgrade_cost_per_sub_php'].astype(float).round(2)
df['est_ftth_upgrade_cost_per_sub_php'] #filtering the content of the conjured col only
```

```python
df.info()
```

```python
# To convert strings to proper case (Title Case), use .str.title() instead of .str.proper()
proper_1 = df['district'].str.title()
proper_1
```

```python
#Loading proper_1 into the df file

df['district'] = proper_1
df
```

```python
#Changing the sub_district col to a Proper Name
proper_2 = df['sub_district'].str.title()
proper_2
```

```python
#Loading proper_2 into df

df['sub_district'] = proper_2
df
```

```python
df['log_date'] = df['log_date'].dt.date
df
```
